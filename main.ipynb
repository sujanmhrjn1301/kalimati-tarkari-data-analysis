Step 1: Load and Audit the Raw Data

```python
import pandas as pd

file_path = "kalimati_tarkari_dataset_merged_cleaned.csv"

df = pd.read_csv(file_path)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nUnits:")
print(df["unit"].value_counts(dropna=False))

print("\nDate range:")
print(df["date"].min(), "to", df["date"].max())

print("\nFirst five rows:")
display(df.head())
```

```python
duplicate_rows = df[df.duplicated(keep=False)].sort_values(
    by=["name", "date"]
)

display(duplicate_rows.head(20))
```

Selecting Targeted Variables

```python
target_vegetables = [
    "tomato",
    "onion",
    "potato",
    "cabbage",
    "cauli",
    "brinjal",
    "chilli",
    "garlic"
]

# Check that all selected vegetables exist in the dataset
available_vegetables = set(df["name"].dropna().unique())

print("Not found:")
print(set(target_vegetables) - available_vegetables)

# Create a separate analysis dataset
analysis_df = df[
    df["name"].isin(target_vegetables) &
    (df["unit"] == "kg")
].copy()

print("Analysis dataset shape:", analysis_df.shape)
display(analysis_df.head())
```

```python
print("Rows per vegetable:")
print(analysis_df["name"].value_counts())

print("\nDate range:")
print(
    analysis_df["date"].min(),
    "to",
    analysis_df["date"].max()
)

print("\nVarieties or types:")
display(
    analysis_df[
        ["name", "type", "color", "location"]
    ].drop_duplicates().sort_values(["name", "type"])
)
```

```python
duplicate_rows = analysis_df[
    analysis_df.duplicated(keep=False)
].sort_values(
    by=["name", "date"]
)

display(duplicate_rows.head(20))
```

```python
# Convert date values into datetime format
analysis_df["date"] = pd.to_datetime(
    analysis_df["date"],
    errors="coerce"
)

# Convert price columns into numeric format
price_columns = ["minimum", "maximum", "average"]

for column in price_columns:
    analysis_df[column] = pd.to_numeric(
        analysis_df[column],
        errors="coerce"
    )

print("Data types after conversion:")
print(analysis_df.dtypes)
```

```python
print("Missing values after conversion:")
print(analysis_df.isna().sum())

print("\nInvalid price relationships:")
display(
    analysis_df[
        analysis_df["minimum"] > analysis_df["maximum"]
    ][["name", "date", "minimum", "maximum", "average"]].head()
)
```

```python
zero_price_rows = analysis_df[
    (analysis_df["minimum"] <= 0) |
    (analysis_df["maximum"] <= 0) |
    (analysis_df["average"] <= 0)
]

print("Rows with zero or negative prices:", len(zero_price_rows))

display(
    zero_price_rows[
        ["name", "date", "type", "minimum", "maximum", "average"]
    ].head(20)
)
```

```python
invalid_average_rows = analysis_df[
    (analysis_df["average"] < analysis_df["minimum"]) |
    (analysis_df["average"] > analysis_df["maximum"])
]

print(
    "Rows where average is outside the minimum-maximum range:",
    len(invalid_average_rows)
)

display(
    invalid_average_rows[
        ["name", "date", "minimum", "maximum", "average"]
    ].head(20)
)
```

```python
# Create time-related features
analysis_df["year"] = analysis_df["date"].dt.year
analysis_df["month"] = analysis_df["date"].dt.month
analysis_df["month_name"] = analysis_df["date"].dt.month_name()

# Create price-related features
analysis_df["price_range"] = (
    analysis_df["maximum"] - analysis_df["minimum"]
)

analysis_df["price_spread_percent"] = (
    analysis_df["price_range"] /
    analysis_df["average"]
) * 100

display(
    analysis_df[
        [
            "name",
            "date",
            "year",
            "month",
            "month_name",
            "minimum",
            "maximum",
            "average",
            "price_range",
            "price_spread_percent"
        ]
    ].head()
)
```

```python
daily_summary = (
    analysis_df
    .groupby(["name", "date"], as_index=False)
    .agg(
        average_price=("average", "median"),
        minimum_price=("minimum", "median"),
        maximum_price=("maximum", "median"),
        price_range=("price_range", "median"),
        price_spread_percent=("price_spread_percent", "median"),
        observations=("average", "size")
    )
)

daily_summary["year"] = daily_summary["date"].dt.year
daily_summary["month"] = daily_summary["date"].dt.month
daily_summary["month_name"] = daily_summary["date"].dt.month_name()

print("Daily summary shape:", daily_summary.shape)

display(daily_summary.head())
```

```python
monthly_summary = (
    daily_summary
    .groupby(["name", "month", "month_name"], as_index=False)
    .agg(
        typical_price=("average_price", "median"),
        mean_price=("average_price", "mean"),
        price_volatility=("average_price", "std"),
        days_observed=("average_price", "count")
    )
    .sort_values(["name", "month"])
)

display(monthly_summary.head(20))
```
